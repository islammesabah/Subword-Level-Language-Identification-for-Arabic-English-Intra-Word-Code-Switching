```python
import os
import csv
import pandas as pd
import numpy as np
```

```python
# load all data
Twitter = [ os.getcwd() + '/Twitter/TwitterData.csv']
Facebook = [ os.getcwd() + '/Facebook/FacebookData.csv']
WhatsApp = [ os.getcwd() + '/WhatsApp/WhatsAppData.csv']
twitter = pd.read_csv(Twitter[0])
facebook = pd.read_csv(Facebook[0])
whatsapp = pd.read_csv(WhatsApp[0])
li = []
li.append(twitter)
li.append(facebook)
li.append(whatsapp)
Data = pd.concat(li,axis=0,ignore_index=True)

# print table statistics
def printTable(Data):
    count = Data.count()[0]
    print(count)
    n_sentence = len(list(set(Data["Sentence#"].values)))
    tag, counts = np.unique(Data["Tag"], return_counts=True)
    tag_mixed, counts_mixed = np.unique(Data["Tag-Mixed"], return_counts=True)

    tags = dict(zip(tag, counts))

    mixed_count = tags["MIXED"]

    unique_tags = Data.drop_duplicates(subset="Token", keep="last")["Tag"].value_counts()
    unique_mixed_tags = Data.drop_duplicates(subset="Token", keep="last")["Tag-Mixed"].value_counts()

    unique_count = 0
    for i in unique_tags:
        unique_count += i

    unique_mixed_count = unique_tags["MIXED"]
    print("{:15}&{:10}&{:10}&{:10}&{:10}".format("Tag", "All", "%","Unique","Unique %"))
    print(60 * "=")
    for tags,counts in zip(tag, counts):
        per_all = "{:.2f}".format( (counts / count)*100.0 )
        unique_t = unique_tags[tags]
        per_unique = "{:.2f}".format((unique_t / unique_count)*100.0)
        print("{:15}&{:10}&{:10}&{:10}&{:10}".format(tags, counts, per_all,unique_t,per_unique))
    print(60 * "=")
    for tags,counts in zip(tag_mixed, counts_mixed):
        if tags not in tag:
            per_all = "{:.2f}".format( (counts / mixed_count)*100.0 )
            unique_t = unique_mixed_tags[tags]
            per_unique = "{:.2f}".format((unique_t / unique_mixed_count)*100.0)
            print("{:15}&{:10}&{:10}&{:10}&{:10}".format(tags, counts, per_all,unique_t,per_unique))

# combine three table to one table and save them and solve the sentences numbers
def combineData(outputName,Data1,Data2,Data3):
    with open(outputName+'.csv', 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['#', 'Sentence#', 'Token', 'Tag', 'Tag-Mixed', 'Segment', 'SegmentLen'])
        j = 1
        sentenceNum = 1
        lastSentenceLen = 1
        for i in range(0,len(Data1)):
            NSentence = Data1['Sentence#'][i]
            if(lastSentenceLen != NSentence):
                sentenceNum += 1
            lastSentenceLen = NSentence
            writer.writerow([j,sentenceNum,Data1['Token'][i],Data1['Tag'][i],Data1['Tag-Mixed'][i],Data1['Segment'][i],Data1['SegmentLen'][i]])
            j+=1
        for i in range(0,len(Data2)):
            NSentence = Data2['Sentence#'][i]
            if(lastSentenceLen != NSentence):
                sentenceNum += 1
            lastSentenceLen = NSentence
            writer.writerow([j,sentenceNum,Data2['Token'][i],Data2['Tag'][i],Data2['Tag-Mixed'][i],Data2['Segment'][i],Data2['SegmentLen'][i]])
            j+=1
        for i in range(0,len(Data3)):
            NSentence = Data3['Sentence#'][i]
            if(lastSentenceLen != NSentence):
                sentenceNum += 1
            lastSentenceLen = NSentence
            writer.writerow([j,sentenceNum,Data3['Token'][i],Data3['Tag'][i],Data3['Tag-Mixed'][i],Data3['Segment'][i],Data3['SegmentLen'][i]])
            j+=1
            
# combine the data and save it to a csv file      
combineData("Data",twitter,facebook,whatsapp)

# print all data table statistics
printTable(Data)
```

```python
# calculate the number of Arabic prefix and suffix in mixed tokens
Data = pd.read_csv(os.getcwd() + '/Data.csv')
def MixedStatistic(Data):
    token = []
    pos = []
    tag = []
    for i in range(0,len(Data)):
        if Data['Tag'][i]=='MIXED':
            segm = Data['Segment'][i].split(',');
            token += segm
            posi = []
            for j in range(1,len(segm)+1):
                posi += [j]
            pos += posi
            tag += Data['Tag-Mixed'][i].split(',')
            
            
    Morph = {'Tokens': token,
             'Tags': tag,
             'Pos': pos
            }
    df = pd.DataFrame(Morph, columns = ['Tokens', 'Tags', 'Pos'])
    return df

Morph = MixedStatistic(Data)
x = Morph.groupby(["Tokens", "Tags","Pos"]).size().reset_index(name="Time")
x.sort_values(by ='Time'  , ascending=False).head(30)
```
