```python
import os
import pandas as pd
import csv
```

```python
# load the data
Messages_path= [ os.getcwd() + '/Messages.csv']
Messages = pd.read_csv(Messages_path[0])
```

```python
# check if there are null message
print(Messages.isnull().any())
```

```python
# get messages length
messages = list(set(Messages["Message"].values))
len(messages)
```

```python
# tokenize the messages
import re
pattern = r'<(!?).*>'
def tokenize(data):
    out = []
    for messages in data:
        # clean the tokens from URL and spaces and unwanted chars
        messages = messages.lower()
        messages = re.sub(r'^https?:\/\/.*[\r\n]*', '', messages, flags=re.MULTILINE)
        messages = re.sub(pattern,'',messages)
        messages = re.sub(r'http\S+', '',  messages)
        l = re.split(r'(\W+)', messages)
        l[:] = (value for value in l if (value != " " and value != ""))
        out.append(l)
    return out
tokens = tokenize(Messages["Message"])
```

```python
# save the tokens with its messages number
with open('Tokeninsentence.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["#","Sentence#","Token"])
    num = 1
    sentenceNum = 1
    for sentence in tokens:
        for token in sentence:
            writer.writerow([num,sentenceNum,token])
            num += 1
        sentenceNum += 1
    file.close()
```

```python
# remove the spaces and the enters and unwanted char in the tokens without segments yet
def removespacewithoutseg(outputName,inputName):
    Data = pd.read_csv(os.getcwd() + '/'+ inputName +'.csv')
    with open(outputName+'.csv', 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['#', 'Sentence#', 'Token', 'Tag', 'Tag-Mixed'])
        j = 1
        for i in range(0,len(Data)):
            token = str(Data['Token'][i]).replace("\n","")
            token = token.replace(" ","")
            token = token.replace("\r","")
            if((token != "") & (token != "nan") and (token != " ")):
                if(str(Data['Tag'][i])!=""):
                    writer.writerow([j,Data['Sentence#'][i],token,Data['Tag'][i],Data['Tag-Mixed'][i]])
                else:
                    writer.writerow([j,Data['Sentence#'][i],token])
                j += 1
                
# remove the spaces and the enters and unwanted char in the tokens with segments                
def removespacewithseg(outputName,inputName):
    Data = pd.read_csv(os.getcwd() + '/'+ inputName +'.csv')
    with open(outputName+'.csv', 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['#', 'Sentence#', 'Token', 'Tag', 'Tag-Mixed', 'Segment', 'SegmentLen'])
        j = 1
        for i in range(0,len(Data)):
            token = Data['Token'][i].replace("\n","")
            token = token.replace(" ","")
            token = token.replace("\r","")
            segment = Data['Segment'][i].replace("\n","")
            segment = segment.replace(" ","")
            segment = segment.replace("\r","")
            if(segment == Data['Segment'][i]):
                segmentLen = Data['SegmentLen'][i]
            else:
                segmentLen = len(segment)
            if((token != "") & (token != " ") & (token != "nan")):
                writer.writerow([j,Data['Sentence#'][i],token,Data['Tag'][i],Data['Tag-Mixed'][i],segment,segmentLen])
                j += 1

#compine two csv                 
def combineTwoData(outputName,inputName1,inputName2):
    Data1 = pd.read_csv(os.getcwd() + '/'+ inputName1 +'.csv')
    Data2 = pd.read_csv(os.getcwd() + '/'+ inputName2 +'.csv')
    with open(outputName+'.csv', 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['#', 'Sentence#', 'Token', 'Tag', 'Tag-Mixed', 'Segment', 'SegmentLen'])
        j = 1
        sentenceNum = 1
        lastSentenceLen = 1
        for i in range(0,len(Data1)):
            NSentence = Data1['Sentence#'][i]
            if(lastSentenceLen != NSentence):
                sentenceNum += 1
            lastSentenceLen = NSentence
            writer.writerow([j,sentenceNum,Data1['Token'][i],Data1['Tag'][i],Data1['Tag-Mixed'][i],Data1['Segment'][i],Data1['SegmentLen'][i]])
            j+=1
        for i in range(0,len(Data2)):
            NSentence = Data2['Sentence#'][i]
            if(lastSentenceLen != NSentence):
                sentenceNum += 1
            lastSentenceLen = NSentence
            writer.writerow([j,sentenceNum,Data2['Token'][i],Data2['Tag'][i],Data2['Tag-Mixed'][i],Data2['Segment'][i],Data2['SegmentLen'][i]])
            j+=1
        
```
