```python
# import the libraries
import tensorflow.compat.v2 as tf
import tensorflow_hub as hub
from tensorflow_text import SentencepieceTokenizer
import glob
```

```python
# upload the pre-trained MUSE embedding model
module_url = 'https://tfhub.dev/google/universal-sentence-encoder-multilingual-large/3' #@param ['https://tfhub.dev/google/universal-sentence-encoder-multilingual/3', 'https://tfhub.dev/google/universal-sentence-encoder-multilingual-large/3']
# You can choose between two models from the right selector

# load the model
model = hub.load(module_url)

# Create a method to embed the inputs
def embed_text(input):
  return model(input)
```

```python
# Import flair to tokenize the sentence
from flair.data import Sentence

# create an example sentence
sentence = Sentence('الحياة gamela gedan we now its agmal .')

# Check the embedding on our example
for token in sentence:
    print(token.text)
    print(embed_text(token.text).numpy().tolist()[0])
```

```python
# load the corpus path
files = glob.glob('/content/gdrive/My Drive/Colab Notebooks/Flair Embedding/corpus/*.*')
files += glob.glob('/content/gdrive/My Drive/Colab Notebooks/Flair Embedding/corpus/*/*')
print(files)
```

```python
# apply the embedding to our corpus tokens
# output file will have multilines every line will take this shape = token_text 1.0 1.1 .... 1.1
from flair.data import Sentence
file_out = open("/content/gdrive/My Drive/Colab Notebooks/Flair Embedding/pre_trained_muse.aren", 'w')
for file in files:
    print(file)
    with open(file, 'r', encoding='utf-8') as f:
        k = 0
        j = 0
        for line in f:
          k = k+1
          sentence = Sentence(line)
          if len(sentence) != 0:
            j = j+1
            for token in sentence:
              output = ""
              output += str(token.text)
              for i in embed_text(token.text).numpy().tolist()[0]:
                output += " "+str(i)
              output += "\n"
              file_out.writelines(output)

file_out.close() 

# show the last token
print(output)
print(k,j)
```
